```python
import numpy as np
import pandas as pd
```

```python
# Sample Data as a dictionary
data = {
    "OrderID": [1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008],
    "Date": ["2025-06-01", "2025-06-02", "2025-06-02", "2025-06-03",
             "2025-06-04", "2025-06-04", "2025-06-05", "2025-06-05"],
    "Customer": ["Asha Rao", "Ravi Kumar", "Neha Patel", "Sunil Verma",
                 "Asha Rao", "Ravi Kumar", "Neha Patel", "Sunil Verma"],
    "Product": ["Notebook", "Ball Pen", "Laptop", "Mouse",
                "Pencil Box", "USB Drive", "Water Bottle", "Notebook"],
    "Category": ["Stationery", "Stationery", "Electronics", "Electronics",
                 "Stationery", "Electronics", "Lifestyle", "Stationery"],
    "Quantity": [3, 5, 1, 2, 1, 4, 2, 5],
    "Price": [50, 10, 55000, 500, 120, 800, 250, 50],
    "City": ["Mumbai", "Bengaluru", "Delhi", "Mumbai",
             "Chennai", "Bengaluru", "Hyderabad", "Chennai"]
}

df = pd.DataFrame(data)

# Convert Date column to datetime
df['Date'] = pd.to_datetime(df['Date'])

print(df.head())
```

```python
# Challenge 1 - Total revenue by product?
df["Revenue"] = df["Quantity"] * df["Price"]
new_var = df[["Category","Revenue"]]
new_var.groupby(new_var["Category"]).sum()["Revenue"]
```

```python
#Optimized code
df["Revenue"] = df["Quantity"] * df["Price"]
x = df.groupby("Category")["Revenue"].sum()
print(x)
```
